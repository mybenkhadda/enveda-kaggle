# 01 — Preprocessing Pipeline

A thin orchestration layer: every reusable step lives in `casmi.*`, this notebook only calls
`run_preprocessing` and inspects the result. Rerunning this notebook is safe and fast --
`run_preprocessing` skips any stage whose artifact is already valid for the current config.


````
from casmi.paths import get_project_paths
from casmi.config import PreprocessingConfig
from casmi.pipelines.preprocessing import run_preprocessing
````

````
paths = get_project_paths()
config = PreprocessingConfig(seed=42, n_jobs=8)
config
````

````
result = run_preprocessing(
    paths=paths,
    config=config,
)
````

## Summary


````
import pandas as pd
pd.Series(result.summary)
````

## Checks


````
checks_df = pd.Series(result.checks).rename("passed").to_frame()
display(checks_df)
assert all(result.checks.values()), "one or more preprocessing checks failed -- see above"
````

## Artifacts


````
for name, path in result.artifacts.items():
    exists = path.exists()
    size_mb = path.stat().st_size / 1e6 if exists else 0.0
    print(f"{name:28s} exists={exists!s:5s}  {size_mb:8.2f} MB  {path}")
````

## Runtime


````
print(f"Total runtime: {result.runtime_seconds:.1f}s")
````
